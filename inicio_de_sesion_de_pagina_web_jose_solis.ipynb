```python
# programa de inicio de sesion para pagina web
usuario_guardado = ""
contrasena_guardada = ""

opcion = 8

while opcion != 3:

   print("----MENU----")
   print("1. registrarse")
   print("2. iniciar sesion")
   print("3. salir")

   opcion = int(input("ingrese una opcion: "))

   if opcion == 1:
      usuario_guardado = input("crea un usuario: ")
      contrasena_guardada = input("crea una contrasena:")

      print("usuario registrado con exito")

   elif opcion == 2:
      usuario_ingresado = input("ingrese su usuario: ")
      contrasena_ingresada = input("ingrese su contrasena: ")

      if usuario_ingresado == usuario_guardado and contrasena_ingresada == contrasena_guardada:
         print("inicio de sesion exitoso")
      else:
         print("usuario o contraseña fallido")

   elif opcion == 3:
        print("saliendo del programa...")
   else:
        print("opcion invalida")
```
